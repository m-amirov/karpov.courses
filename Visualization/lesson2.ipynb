```python
import pandas as pd
```

```python
products = pd.read_excel("data1.xlsx", sheet_name=1)
dynamics = pd.read_excel("data1.xlsx", sheet_name=0)
```

Сколько уникальных Брендов представлено в нашем магазине?

```python
products['Бренд'].nunique()
```

У какого бренда больше всего уникальных SKU?

```python
products.groupby('Бренд', as_index=False).agg(n_SKU=('SKU', 'nunique')).sort_values(by='n_SKU', ascending=False).head(1)
```

За какой период времени у нас есть данные о выручке и продажах по SKU?

```python
print(dynamics.sort_values(by='Месяц', ascending=True).head(1)['Месяц'])
print(dynamics.sort_values(by='Месяц', ascending=False).head(1)['Месяц'])
```

Сколько уникальных товарных позиций (SKU) есть в вашем датасете?

```python
products2['SKU'].nunique()
```

Сколько различных категорий товаров у бренда AVON?



```python
products.groupby('Бренд', as_index=False).agg(n_SKU=('Категория', 'nunique')).query('Бренд == "AVON"')
```

```python
products.query('Страна == "Республика Корея"')['Бренд'].nunique()
```

Для анализа финансовых показателей у нас есть две таблицы: Dynamic и Product. Чтобы построить аналитику, нужно будет объединить данные. Какой столбец используем для JOIN?

```python
df = products.merge(dynamics, on='Бренд', how='inner')
```
