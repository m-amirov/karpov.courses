Кейсы

Вы — аналитик продукта EasyRead, популярного приложения для чтения книг. EasyRead позволяет пользователям читать книги онлайн, слушать аудиокниги и находить рекомендации по книгам на основе их интересов.


Команда продукта хочет улучшить ключевую метрику — конверсию в подписчиков (платящих клиентов). На текущий момент:


Всего у приложения 12 млн пользователей, из них 3 млн — платящие подписчики (конверсия: 25%).

Для упрощения в этом задании будем считать, что количество новых привлеченных пользователей равно количеству ушедших пользователей.

Продакт-менеджер сформировал 5 гипотез, которые могут улучшить конверсию. Он предоставил вводные данные о каждой гипотезе:


Результаты предварительных исследований.

Примерное время и ресурсы для реализации.

Ваша задача — рассчитать, какая гипотеза имеет наивысший приоритет, используя методику ICE.


Список гипотез:

Гипотеза 1. Рекомендации

​​Если внедрим блок персонализированных рекомендаций, то сегменту подростков (20% неплатящих пользователей) не придется самостоятельно искать интересные книги, вследствие чего конверсия в подписчиков увеличится.


Impact: Ожидается, что 30% из сегмента подростков начнут конвертироваться в подписчиков.

Confidence: фича с рекомендациями есть у всех конкурентов.

Ease: Реализация требует работы 2 разработчиков и 1 аналитика в течение 2 недель.

Гипотеза 2. Ускорение загрузки

Если ускорим загрузку страниц книг, то пользователям с медленным интернетом не придется ждать, пока данные загрузятся, вследствие чего конверсия в подписчиков увеличится.


Impact: Изменения затронут 30% всех пользователей, увеличив их конверсию на 2 п.п.

Confidence: среди клиентов есть запрос на этот функционал, у 1 ключевого конкурента есть такая фича, однако никаких исследований в рамках нашей компании - нет.

Ease: Требуется 3 недели работы команды backend-разработчиков.

Гипотеза 3. Тексты в email-рассылках

Если внедрим более персонализированные тексты в email-рассылках, то пользователи будут меньше игнорировать письма, вследствие чего метрика конверсии в подписчиков увеличится.


Impact: Конверсия увеличится на 5% среди всех пользователей (подразумевается относительное увеличение).

Confidence: A/B-тесты показали, что персонализация увеличивает открытие писем на 10%, но прирост подписок не тестировался.

Ease: Требуется 3 дня работы копирайтера.

Гипотеза 4. Пробный период с расширенным функционалом

Если внедрим пробный период с расширенным функционалом для новых пользователей, то они смогут лучше оценить ценность сервиса, вследствие чего количество конверсий в платящих пользователей увеличится.


Impact: Пробный период с расширенным функционалом увеличит конверсию в платящих на 10% (подразумевается относительное увеличение).

Confidence: Делали запуск крупномасштабного MVP, данные после запуска подтверждали гипотезу.

Ease: Для реализации потребуется 1 неделя на доработку системы и настройки для разных тарифных планов.

Гипотеза 5. Тариф по сниженной цене

Если введём тариф с базовыми функциями по сниженной цене, то пользователи с низким бюджетом смогут позволить себе подписку, вследствие чего общее количество подписчиков увеличится.


Impact: Привлечёт 5% неплатящих пользователей.

Confidence: Продакт верит в эту гипотезу и внешний эксперт подтвердил, что гипотеза должна сработать.

Ease: Потребуется 1 месяц работы всех команд: маркетинг, аналитика, разработка.
