```python
import pandas as pd
import pingouin as pg
import statsmodels.api as sm
```

```python
df = pd.read_csv("../Statistic/data/cars.csv")
```

```python
df['company'] = df.CarName.apply(lambda x: x.split(' ')[0])
```

```python
df = df.drop(columns=['CarName', 'car_ID'])
```

```python
df['company'] = df.company.replace(to_replace={'maxda':'mazda', 
                                               'Nissan':'nissan',
                                               'porcshce':'porsche',
                                               'toyouta':'toyota',
                                               'vokswagen':'volkswagen',
                                               'vw':'volkswagen'
                                               })
```

```python
df['company'].nunique()
```

```python
cols_to_keep = [['price'], ['company', 'fueltype', 'aspiration','carbody', 
'drivewheel', 'wheelbase', 'carlength','carwidth', 
'curbweight', 'enginetype', 'cylindernumber', 'enginesize', 
'boreratio','horsepower']]

cols_to_keep1 = ['company', 'fueltype', 'aspiration','carbody', 
'drivewheel', 'wheelbase', 'carlength','carwidth', 
'curbweight', 'enginetype', 'cylindernumber', 'enginesize', 
'boreratio','horsepower', 'price']

df = df[cols_to_keep1]
```

```python
pg.pairwise_corr(data=df, columns=cols_to_keep)
```

```python
df_dummy = pd.get_dummies(data=df, drop_first=True)
for col in df_dummy.select_dtypes(include='bool').columns:
    df_dummy[col] = df_dummy[col].map({True: 1, False: 0})
```

```python
import statsmodels.formula.api as smf
formula = "price ~ horsepower"

smf.ols(formula, df).fit().summary()
```

```python
df_dummy = pd.get_dummies(data=df, drop_first=True)
for col in df_dummy.select_dtypes(include='bool').columns:
    df_dummy[col] = df_dummy[col].map({True: 1, False: 0})
X = df_dummy.drop(['price'], axis='columns') #предикторы
y = df_dummy['price'] #целевая переменная

#не забываем про константу
X = sm.add_constant(X)

sm.OLS(y, X).fit().summary()
```

```python
cols_to_keep = df_dummy.columns[~df_dummy.columns.str.startswith('company_')]
df_dummy_no_company = df_dummy[cols_to_keep]

X = df_dummy_no_company.drop(['price'], axis='columns') #предикторы
y = df_dummy_no_company['price'] #целевая переменная

#не забываем про константу
X = sm.add_constant(X)

sm.OLS(y, X).fit().summary()
```
